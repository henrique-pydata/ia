```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv('../data/processed/titanic_processed.csv')
```

```python
df.info()
```

```python
df.describe()
```

```python
cat_cols = ['survived', 'sex', 'title', 'deck', 'ticket_group_size', 'pclass']
```

```python
fig, axes = plt.subplots(3, 2, figsize=(12, 8))

for i, c in enumerate(cat_cols): 
    sns.countplot(data=df, x=c, hue=c, ax=axes.flat[i])

plt.tight_layout()
plt.show()
```

```python
round((df.groupby('ticket_group_size')['survived'].mean()) * 100, 2)
```

```python
female = df[df['sex'] == 'female']
male = df[df['sex'] == 'male']

sns.countplot(data = female, x='sex', hue='survived')
sns.countplot(data = male, x='sex', hue='survived')

plt.show()
```
